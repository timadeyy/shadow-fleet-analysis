```python
import pandas as pd
import os
from utils import get_vessel_info, parse_vessel_data

vessels = pd.read_csv('../data/shadow_fleet_vessels.csv')
```

```python
if os.path.exists('../data/vessel_summary.csv'):
    summary_df = pd.read_csv('../data/vessel_summary.csv')
    identities_df = pd.read_csv('../data/vessel_identities.csv')
    print("Data loaded from cache")
else:
    all_summaries = []
    all_identities_full = []
    for imo in vessels['imo']:
        raw = get_vessel_info(imo)
        summary, identities = parse_vessel_data(imo, raw)
        all_summaries.append(summary)
        all_identities_full.extend(identities)

    summary_df = pd.DataFrame(all_summaries)
    identities_df = pd.DataFrame(all_identities_full)
    summary_df.to_csv('../data/vessel_summary.csv', index=False)
    identities_df.to_csv('../data/vessel_identities.csv', index=False)
    print("Collected by API and saved in cache")
```
