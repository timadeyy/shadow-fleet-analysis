# Where are shadow fleet tankers managed from?

**Question:** In which countries are the ISM managers of the 50 tankers based?

**Why ISM manager:** the registered owner is often a shell company in an offshore
jurisdiction, while the ISM manager is responsible for the ship's daily operation
and safety, so it is closer to the real operator.

```python
import pandas as pd
shadow_FV = pd.read_csv('../data/shadow_fleet_vessels.csv')
```

```python
shadow_FV[['registered_owner', 'ship_manager', 'ism_manager']].head(10)
```

```python
#Sorting places for easier comparison
shadow_FV['ism_country'] = shadow_FV['ism_manager'].str.split('.').str[0]

shadow_FV['reg_country'] = shadow_FV['registered_owner'].str.split('.').str[0]

shadow_FV['manager_country'] = shadow_FV['ship_manager'].str.split('.').str[0]

```

```python
columns = [ 'ism_country', 'reg_country',  'manager_country']
counts = {}
for col in columns:
    counts[col] = shadow_FV[col].value_counts()

comparision_df = pd.DataFrame(counts).fillna(0).astype(int)
comparision_df.sort_values(by = ['reg_country'], ascending = False)
```

```python
#For how many of the 50 ships is the owner country different from the ISM country?
shadow_FV['owners_ism_mismatch'] =  shadow_FV['reg_country'] != shadow_FV['ism_country']
shadow_FV['owners_ism_mismatch'].head(10)
```

```python
pd.crosstab(shadow_FV['reg_country'], shadow_FV['ism_country'])
```

```python
comparision_df['net'] = comparision_df['ism_country'] - comparision_df['reg_country']
comparision_df.sort_values('net', ascending = False)
```
